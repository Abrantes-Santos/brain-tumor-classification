# 🧠 Classificação de Tumores Cerebrais em Ressonância Magnética (MRI) via PyTorch

Pipeline em PyTorch para classificação de tumores cerebrais em 4 classes clínicas:
- **Glioma (`glioma_tumor`)**
- **Meningioma (`meningioma_tumor`)**
- **Adenoma Hipofisário (`pituitary_tumor`)**
- **Sem Tumor (`no_tumor`)**

## 1. Importação dos Módulos e Configuração de Ambiente

```python
import sys
import os

# Adiciona a pasta raiz ao PATH para importar da pasta src
sys.path.append(os.path.abspath(".."))

from src.utils import set_seed, get_device, plot_batch_samples, plot_learning_curves, plot_prediction
from src.dataset import build_dataloaders
from src.model import BrainTumorCNN, count_parameters
from src.train import train

set_seed(42)
device = get_device()
```

## 2. Carregamento dos Dados (DataLoaders)

```python
from google.colab import drive
drive.mount('/content/drive')

DATA_PATH = "/content/drive/MyDrive/Training"
IMG_SIZE = 128
BATCH_SIZE = 32

train_loader, val_loader, class_names = build_dataloaders(
    data_path=DATA_PATH,
    img_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    train_split=0.8,
    seed=42
)

print(f"Classes detectadas: {class_names}")
```

## 3. Visualização de Amostras

```python
plot_batch_samples(train_loader, class_names=class_names, num_samples=8)
```

## 4. Instanciação do Modelo

```python
model = BrainTumorCNN(num_classes=len(class_names), dropout_rate=0.4).to(device)
print(model)
print(f"\nTotal de Parâmetros Treináveis: {count_parameters(model):,}")
```

## 5. Treinamento e Curvas de Desempenho

```python
import argparse

args = argparse.Namespace(
    data_path=DATA_PATH,
    save_dir="../checkpoints",
    epochs=40,
    batch_size=BATCH_SIZE,
    img_size=IMG_SIZE,
    lr=0.001,
    dropout=0.4,
    train_split=0.8,
    num_workers=2,
    seed=42
)

history = train(args)
plot_learning_curves(history, save_path="../reports/figures/learning_curves.png")
```
